```python
posts = [
"Arrancando el lunes con energía #Motivación #NuevaSemana",
"Terminé mi primer proyecto en Python #Python #Programación #OrgullosoDeMi",
"No puedo creer el final de la serie #SinSpoilers #SerieAdicta",
"Nuevo video en el canal sobre #InteligenciaArtificial y #Python",
"Entrenamiento de hoy completado #Fitness #Motivación #NoPainNoGain",
"Leyendo sobre #InteligenciaArtificial y el futuro deltrabajo #Tecnología",
"Arranqué a estudiar #Programación por mi cuenta #Python #Autodidacta",
"Finde de lluvia, maratón de series #SerieAdicta #Relax",
"Workshop de #InteligenciaArtificial en la universidad #Tecnología #Programación"
]
```

```python
new_posts = " ".join(posts).split()
hashtags = {}

for word in new_posts:
  if word.startswith("#"):
    hashtags[word] = 1 if word not in hashtags else hashtags[word] + 1



filter_hashtags = {k: v for k,v in hashtags.items() if v > 1}
sorted_hashtags = sorted(filter_hashtags.items(), key = lambda item: item[1], reverse=True)
unique_hashtags = len(hashtags)


print(f"Hashtags trending (más de una aparición): \n")
for tag, count in sorted_hashtags:
    print(f"  {tag}: {count}")

  
print(f"\nTotal de hashtags únicos: {unique_hashtags}")
```
