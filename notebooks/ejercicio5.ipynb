```python
calculating = {
    "hasta 1kg" : { 
        "local" : 500,
        "regional" : 1000,
        "nacional" : 2000
    },
    "Entre 1 y 5 kg" : {
        "local" : 1000,
        "regional" : 2500,
        "nacional" : 4500
    },
    "Más de 5 kg" : {
        "local" : 2000,
        "regional" : 5000,
        "nacional" : 8000
    }
}
```

```python
weight = float(input("Ingrese el peso del paquete (kg): "))
zone = str(input("Ingrese la zona de destino (local/regional/nacional): ")).lower()

if zone == "local" or zone == "regional" or zone == "nacional":
  if weight <= 1:
    valor = calculating["hasta 1kg"][zone]
    print(f"Costo de envío: ${valor}")
  elif weight <= 5:
    valor = calculating["Entre 1 y 5 kg"][zone]
    print(f"Costo de envío: ${valor}")
  else:
    valor = calculating["Más de 5 kg"][zone]
    print(f"Costo de envío: ${valor}")
else:
  print("Zona no válida. Las zonas disponibles son: local, regional, nacional.")
```
