```python
email = input("Ingrese una dirección de email: ")

if email.startswith("@") or email.startswith(".") or email.endswith("@") or email.endswith(".") or email.count("@") != 1 or not "." in email:
  print("El email no es válido")
else:
  index_a = email.rfind("@")
  index_p = email.rfind(".")

  if index_p < index_a:
    print("El email no es válido")
  else:
    if len(email[index_p+1:]) >= 2:
      print("El email es válido")
    else:
      print("El email no es válido")
```
