```python
player_statistics = {
    'Valentina': {
        'score': 0, 'wins': 0, 'best': 0, 'mean': 0},

    'Mateo': {
        'score': 0, 'wins': 0, 'best': 0, 'mean': 0},

    'Camila': {
        'score': 0, 'wins': 0, 'best': 0, 'mean': 0},

    'Santiago': {
        'score': 0, 'wins': 0, 'best': 0, 'mean': 0},

    'Lucía': {
        'score': 0, 'wins': 0, 'best': 0, 'mean': 0},
}
```

```python
round_win = ''
round_win_score = 0

for i, round in enumerate(rounds):
  for player in round['scores']:
    round_score = round['scores'][player].values()
    total_round_score = sum(round_score)
    player_statistics[player]['score'] += total_round_score

    if player_statistics[player]['best'] < total_round_score:
      player_statistics[player]['best'] = total_round_score

    if total_round_score > round_win_score:
      round_win_score = total_round_score
      round_win = player

  player_statistics[round_win]['wins'] += 1


  print(f"Ronda {i + 1} - {round['theme']}: \n Ganador: {round_win} ({round_win_score})")
  print(f".... ")
  table_statistics(player_statistics, 'tabla de posiciones')
  round_win_score = 0

for player in player_statistics:
  score = player_statistics[player]['score']
  player_statistics[player]['mean'] = score / len(rounds)


table_statistics(player_statistics, 'Tabla de posiciones final')
```
